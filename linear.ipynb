```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Dataset
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Hours": [1, 2, 3, 4, 5],
    "Attendance": [60, 65, 70, 75, 80],
    "Marks": [35, 40, 50, 55, 65]
}

# 2. DataFrame
df = pd.DataFrame(data)

# 3. Input and Target
X = df["Hours"].values
y = df["Marks"].values

# 4. Initial values
m = 0
c = 0

# 5. Learning rate
learning_rate = 0.01

# 6. Number of iterations
iterations = 1000

# 7. Number of data points
n = len(X)

# 8. Gradient Descent
for i in range(iterations):

    # Prediction
    y_pred = m * X + c

    # Error
    error = y_pred - y

    # Gradients
    dm = (2/n) * np.sum(X * error)
    dc = (2/n) * np.sum(error)

    # Update m and c
    m = m - learning_rate * dm
    c = c - learning_rate * dc

# 9. Display model parameters
print("Slope:", m)
print("Intercept:", c)

# 10. Predict new student's marks
new_hours = 6
prediction = m * new_hours + c

print("Predicted Marks:", prediction)

# 11. Regression line
y_pred = m * X + c

# 12. Graph
plt.scatter(X, y)
plt.plot(X, y_pred)

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression using Gradient Descent")

plt.show()
```
